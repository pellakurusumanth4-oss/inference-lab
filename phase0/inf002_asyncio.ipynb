```python
import time
from typing import Iterator

def token_stream(text:str,delay:float=0.6)->Iterator[str]:
  for word in text.split():
    time.sleep(delay)
    yield word
if __name__ == "__main__":
  start=time.perf_counter()
  for i,tok in enumerate(token_stream("The quick brown fix jumps over the lazy dog")):
    if i ==0:
      print(f"[TTFT:{time.perf_counter()-start:.2f}s]")
    print(tok,end=" ",flush=True)
  print(f"\n[total:{time.perf_counter()-start:.2f}s]")
```

```python
import time
from typing import Iterator

def token_stream(text: str, delay: float = 0.1) -> Iterator[str]:
    for word in text.split():
        time.sleep(delay)
        yield word

if __name__ == "__main__":
    for tok in token_stream("The quick brown fox"):
        print(tok, end=" ", flush=True)
    print()
```

```python
import asyncio
import time

async def fake_requests(req_id: int, latency: float) -> str:
    print(f"req {req_id} started")
    await asyncio.sleep(latency)
    print(f"req {req_id} done after {latency}s")
    return f"result-{req_id}"

async def main() -> None:
    latencies: list[float] = [1.0, 2.0, 1.5]
    start = time.perf_counter()
    results: list[str] = await asyncio.gather(
        *(fake_requests(i, lat) for i, lat in enumerate(latencies))
    )
    print(f"Concurrent: {time.perf_counter() - start:.2f}s")
    print(results)

if __name__ == "__main__":
    # In Jupyter/Colab notebooks, use await directly instead of asyncio.run()
    await main()
```
